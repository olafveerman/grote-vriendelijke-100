# GV100: first look
Starter analyses on `data/`. 2022 is partial, so most comparisons use the four complete editions.

```python
import pandas as pd
long = pd.read_csv('../data/gv100_long.csv')
wide = pd.read_csv('../data/gv100_wide.csv')
FULL = [2021, 2023, 2024, 2025]
long.groupby('year').size()
```

## Evergreens: books in every complete edition

```python
cols = [f'rank_{y}' for y in FULL]
ever = wide.dropna(subset=cols).copy()
ever['mean_rank'] = ever[cols].mean(axis=1)
print(len(ever), 'books in all four complete editions')
ever.sort_values('mean_rank')[['title','author']+cols+['mean_rank']].head(20)
```

## Turnover: new entries and drop-outs between consecutive complete editions

```python
sets = {y: set(long.loc[long.year==y,'work_key']) for y in FULL}
for a, b in zip(FULL, FULL[1:]):
    print(f'{a}->{b}: {len(sets[b]-sets[a])} new, {len(sets[a]-sets[b])} dropped')
```

## Most-listed authors per edition

```python
(long[long.year.isin(FULL)]
 .groupby(['author','year']).size().unstack(fill_value=0)
 .assign(total=lambda d: d.sum(axis=1))
 .sort_values('total', ascending=False).head(15))
```

## Biggest movers 2024 → 2025

```python
m = wide.dropna(subset=['rank_2024','rank_2025']).copy()
m['change'] = m.rank_2024 - m.rank_2025
m.sort_values('change')[['title','author','rank_2024','rank_2025','change']].iloc[list(range(5))+list(range(-5,0))]
```
